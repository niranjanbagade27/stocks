```python
import pandas as pd
import os
import shutil

# --- Configuration ---
# Define the name of the folder containing all the company folders.
# This folder should be in the same directory as the script.
stocks_folder_name = 'stocks'

# Define the name of the output directory where the combined files will be saved.
output_folder_name = '10YearClubbedData'

# Get the path to the current working directory
base_path = os.getcwd()

# Construct the full paths for the stocks and output folders
stocks_folder_path = os.path.join(base_path, stocks_folder_name)
output_folder_path = os.path.join(base_path, output_folder_name)

# --- Main Script ---

# Check if the main stocks folder exists
if not os.path.exists(stocks_folder_path):
    print(f"Error: The main folder '{stocks_folder_path}' was not found. Please ensure it exists.")
else:
    # Delete the output folder if it exists, to ensure a clean run.
    if os.path.exists(output_folder_path):
        shutil.rmtree(output_folder_path)
        print(f"Deleted existing output folder: {output_folder_path}")

    # Create the new output folder
    os.makedirs(output_folder_path)
    print(f"Created new output folder: {output_folder_path}")

    # Get a list of all company folders inside the stocks folder
    company_folders = [f for f in os.listdir(stocks_folder_path) if os.path.isdir(os.path.join(stocks_folder_path, f))]

    if not company_folders:
        print(f"No company folders found inside '{stocks_folder_path}'.")
    else:
        # Loop through each company folder found
        for folder_name in company_folders:
            print(f"\n--- Processing folder: {folder_name} ---")
            
            # Create an empty list to hold the dataframes from each file
            all_dataframes = []

            # Construct the full path to the current company folder
            current_company_folder_path = os.path.join(stocks_folder_path, folder_name)
            
            try:
                for file in os.listdir(current_company_folder_path):
                    # Check if the file is a CSV and not a hidden file
                    if file.endswith('.csv') and not file.startswith('.'):
                        # Construct the full file path
                        file_path = os.path.join(current_company_folder_path, file)
                        print(f"  Reading {file_path}...")
                        
                        try:
                            # Read the CSV file into a pandas DataFrame
                            df = pd.read_csv(file_path)
                            
                            # Remove leading/trailing spaces from column names
                            df.columns = df.columns.str.strip()
                            
                            # Append the DataFrame to our list
                            all_dataframes.append(df)
                        except Exception as e:
                            print(f"  Error reading {file_path}: {e}")
            except Exception as e:
                print(f"An unexpected error occurred while processing {folder_name}: {e}")
                continue # Skip to the next folder in the list

            # Check if any data was read from the current folder
            if all_dataframes:
                # Concatenate all dataframes into a single dataframe
                combined_df = pd.concat(all_dataframes, ignore_index=True)
                
                # Sort the combined DataFrame by the 'Date' column in ascending order.
                # First, convert the column to datetime objects to ensure correct sorting.
                combined_df['Date'] = pd.to_datetime(combined_df['Date'])
                combined_df.sort_values(by='Date', inplace=True)
                
                # Define a dynamic output file name based on the folder name
                output_file_name = f'{folder_name}-Combined-Data.csv'
                output_file_path = os.path.join(output_folder_path, output_file_name)
                
                # Save the combined dataframe to a new CSV file inside the output directory
                # `index=False` prevents pandas from writing the DataFrame index to the CSV
                combined_df.to_csv(output_file_path, index=False)
                
                print(f"\nSuccessfully combined all files into '{output_file_path}'!")
                print(f"Total rows in combined file: {len(combined_df)}")
            else:
                print(f"\nNo CSV files were found in the '{folder_name}' folder. Skipping...")

print("\n--- All folders have been processed ---")
```
